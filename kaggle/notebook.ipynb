*Sample data is synthetic and for demonstration only.*

```python
import glob
import pandas as pd

# Finds the CSV wherever Kaggle mounts the dataset
path = glob.glob('/kaggle/input/**/stablecoin_payment_benchmarks.csv', recursive=True)[0]
df = pd.read_csv(path)
df.head()
```

## Payment status by network

```python
df.groupby(['network','status']).size().unstack(fill_value=0)
```

## Why payments failed

```python
df[df['status'] == 'failed'][['payment_id','network','reason']]
```

## Confirmation progress

```python
df['confirmation_pct'] = (df['confirmations'] / df['required_confirmations'] * 100).clip(upper=100).round(0)
df[['payment_id','network','confirmations','required_confirmations','confirmation_pct','status']]
```

## Amount difference (observed − expected)

```python
df['amount_diff'] = (df['tx_amount'] - df['expected_amount']).round(2)
df[df['amount_diff'].notna() & (df['amount_diff'] != 0)][['payment_id','expected_amount','tx_amount','amount_diff','status']]
```
